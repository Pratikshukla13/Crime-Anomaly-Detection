```python
import torch
import cv2
import numpy as np
from torchvision import transforms
#from network.anomaly_detector_model import AnomalyDetector
from network.c3d import C3D


# Load models
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Load C3D model
c3d_model = C3D(pretrained=False).to(device)
state_dict = torch.load('../pretrained_models/c3d_epoch_80000_state_dict.pt', map_location=device)
c3d_model.load_state_dict(state_dict, strict=False)
c3d_model.eval()

# Load Anomaly Detector
torch.serialization.add_safe_globals([AnomalyDetector])
anomaly_model = torch.load('../pretrained_models/c3d_epoch_80000.pt', map_location=device, weights_only=False)
anomaly_model.eval()

# Preprocessing
transform = transforms.Compose([
    transforms.Resize((112, 112)),
    transforms.ToTensor(),
    transforms.Normalize([0.5], [0.5])
])

def preprocess_clip(clip):
    # clip shape: (16, H, W, 3)
    frames = []
    for frame in clip:
        img = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        img = cv2.resize(img, (112, 112))
        img = transform(img)
        frames.append(img)
    video_tensor = torch.stack(frames).permute(1, 0, 2, 3)  # (C, D, H, W)
    return video_tensor.unsqueeze(0).to(device)

def analyze_video(video_path):
    cap = cv2.VideoCapture(video_path)
    clip = []
    fps = cap.get(cv2.CAP_PROP_FPS)
    frame_count = 0

    while True:
        ret, frame = cap.read()
        if not ret:
            break

        clip.append(frame)
        frame_count += 1

        if len(clip) == 16:
            input_tensor = preprocess_clip(clip)

            with torch.no_grad():
                c3d_features = c3d_model(input_tensor)
                prediction = anomaly_model(c3d_features)

            predicted_class = torch.argmax(prediction, dim=1).item()
            timestamp = frame_count / fps

            print(f"🕒 Time: {timestamp:.2f}s — 🔍 Detected Class: {predicted_class}")
            clip = []

    cap.release()

# 🔽 Replace this path with your input video
analyze_video("test_video.mp4")
```

```python
import cv2
import torch
import numpy as np
from network.c3d import C3D
from network.anomaly_detector_model import AnomalyDetector
from torchvision import transforms

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Load models
c3d = C3D(pretrained=False).to(device)
c3d.load_state_dict(torch.load('../pretrained_models/c3d_epoch_80000_state_dict.pt', map_location=device))
c3d.eval()

anomaly_model = AnomalyDetector().to(device)
anomaly_model.load_state_dict(torch.load('../pretrained_models/anomaly_detector_weights.pt', map_location=device))  # Your saved model
anomaly_model.eval()

# Preprocessing
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Resize((112, 112)),
    transforms.Normalize([0.5], [0.5])  # Adjust if needed
])

def extract_clips(video_path, clip_len=16, step=8):
    cap = cv2.VideoCapture(video_path)
    frames = []
    clips = []

    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break
        frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        frames.append(frame)
        
        if len(frames) >= clip_len:
            clip = frames[-clip_len:]
            clip_tensor = torch.stack([transform(f).unsqueeze(0).squeeze(0) for f in clip])  # (16, 3, 112, 112)
            clips.append(clip_tensor)
            frames = frames[step:]  # Overlap control

    cap.release()
    return clips

def predict_anomalies(video_path, threshold=0.5):
    clips = extract_clips(video_path)
    results = []

    for i, clip in enumerate(clips):
        clip = clip.permute(1, 0, 2, 3).unsqueeze(0).to(device)  # (1, 3, 16, 112, 112)
        with torch.no_grad():
            features = c3d(clip)
            score = anomaly_model(features).item()
            label = "Anomalous" if score >= threshold else "Normal"
            results.append((i, score, label))
            print(f"Clip {i}: Score={score:.3f}, Label={label}")
    
    return results
```

```python
import torch
import torch.nn as nn
import cv2
import numpy as np
from torchvision import transforms
from network.c3d import C3D
from network.anomaly_detector_model import AnomalyDetector

# ✅ Define the crime classifier
class CrimeClassifier(nn.Module):
    def __init__(self, input_dim=4096, num_classes=4):
        super(CrimeClassifier, self).__init__()
        self.classifier = nn.Sequential(
            nn.Linear(input_dim, 512),
            nn.ReLU(),
            nn.Dropout(0.5),
            nn.Linear(512, num_classes)
        )

    def forward(self, x):
        return self.classifier(x)

# ✅ Crime label mapping
crime_labels = ["abuse", "assault", "arson", "arrest"]

# ✅ Preprocessing transform
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Resize((112, 112)),
])

# ✅ Device setup
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# ✅ Load C3D with safe partial weights
c3d = C3D(pretrained=False).to(device)
checkpoint_path = "../pretrained_models/c3d_epoch_80000_state_dict.pt"
checkpoint = torch.load(checkpoint_path, map_location=device)
model_dict = c3d.state_dict()
filtered_dict = {k: v for k, v in checkpoint.items() if k in model_dict and v.size() == model_dict[k].size()}
model_dict.update(filtered_dict)
c3d.load_state_dict(model_dict, strict=False)
c3d.eval()

# ✅ Load crime classifier
classifier = CrimeClassifier().to(device)
classifier.load_state_dict(torch.load("../pretrained_models/crime_classifier.pt", map_location=device))
classifier.eval()

# ✅ Function to read and preprocess video
def load_video_frames(video_path, max_frames=16):
    cap = cv2.VideoCapture(video_path)
    frames = []
    while len(frames) < max_frames and cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break
        frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        frame = transform(frame)
        frames.append(frame)
    cap.release()
    
    if len(frames) < max_frames:
        frames += [frames[-1]] * (max_frames - len(frames))  # pad
    video_tensor = torch.stack(frames).permute(1, 0, 2, 3)  # (C, D, H, W)
    return video_tensor.unsqueeze(0)  # (1, C, D, H, W)

# ✅ Run inference
def predict_crime(video_path):
    video_tensor = load_video_frames(video_path).to(device)
    with torch.no_grad():
        features = c3d(video_tensor)  # (1, 4096)
        outputs = classifier(features)  # (1, 4)
        predicted_class = torch.argmax(outputs, dim=1).item()
        print("✅ Predicted Crime:", crime_labels[predicted_class])

# ✅ Example usage
predict_crime("../test_video.mp4")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler
from sklearn.metrics import classification_report
import numpy as np
import os
import pickle
from tqdm import tqdm
from classifier import CrimeClassifier

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# ✅ Custom Dataset
class C3DFeaturesDataset(Dataset):
    def __init__(self, data, labels):
        self.data = torch.tensor(data, dtype=torch.float32)
        self.labels = torch.tensor(labels, dtype=torch.long)

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        return self.data[idx], self.labels[idx]

# ✅ Load your saved C3D features and labels
with open('../features/features.pkl', 'rb') as f:
    data_dict = pickle.load(f)

train_features, train_labels = data_dict['train_features'], data_dict['train_labels']
test_features, test_labels = data_dict['test_features'], data_dict['test_labels']

# ✅ Dataset & balancing
train_dataset = C3DFeaturesDataset(train_features, train_labels)
test_dataset = C3DFeaturesDataset(test_features, test_labels)

# Handle class imbalance
class_counts = np.bincount(train_labels)
weights = 1. / class_counts
sample_weights = [weights[label] for label in train_labels]
sampler = WeightedRandomSampler(weights=sample_weights, num_samples=len(train_labels), replacement=True)

train_loader = DataLoader(train_dataset, batch_size=16, sampler=sampler)
test_loader = DataLoader(test_dataset, batch_size=16, shuffle=False)

# ✅ Model, Loss, Optimizer
model = CrimeClassifier().to(device)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=1e-4)

# ✅ Training Loop
num_epochs = 30
for epoch in range(num_epochs):
    model.train()
    epoch_loss = 0
    for inputs, targets in tqdm(train_loader, desc=f"Epoch {epoch+1}/{num_epochs}"):
        inputs, targets = inputs.to(device), targets.to(device)

        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, targets)
        loss.backward()
        optimizer.step()

        epoch_loss += loss.item()

    # ✅ Test accuracy
    model.eval()
    correct, total = 0, 0
    all_preds, all_labels = [], []
    with torch.no_grad():
        for inputs, targets in test_loader:
            inputs, targets = inputs.to(device), targets.to(device)
            outputs = model(inputs)
            _, preds = torch.max(outputs, 1)
            correct += (preds == targets).sum().item()
            total += targets.size(0)
            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(targets.cpu().numpy())

    print(f"Epoch {epoch+1}/{num_epochs} - Loss: {epoch_loss:.4f} - Test Accuracy: {correct / total:.2f}")

# ✅ Report & Save
print("\nClassification Report:")
print(classification_report(all_labels, all_preds, target_names=['abuse', 'assault', 'arson', 'arrest']))

os.makedirs('../pretrained_models', exist_ok=True)
torch.save(model.state_dict(), '../pretrained_models/crime_classifier.pt')
print("\n✅ Classifier saved at: ../pretrained_models/crime_classifier.pt")
```

```python
import os
import torch
import torch.nn as nn
import cv2
import numpy as np
import pickle
from tqdm import tqdm
import sys
sys.path.append('./anomaly_detection/notebooks/classifier.py')  # or path to classifier.py
from classifier import C3D
  # Make sure this is your custom C3D class
from torchvision import transforms

# ✅ Set device
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

# ✅ Define video transform
transform = transforms.Compose([
    transforms.ToPILImage(),
    transforms.Resize((112, 112)),
    transforms.ToTensor()
])

# ✅ Load pretrained C3D model (without classifier layers)
class FeatureExtractor(nn.Module):
    def __init__(self, model):
        super().__init__()
        self.features = nn.Sequential(*list(model.children())[:-1])  # Remove classifier
    def forward(self, x):
        return self.features(x)

c3d_model = C3D().to(device)
c3d_model.load_state_dict(torch.load("../pretrained_models/c3d_epoch_80000_state_dict.pt", map_location=device))
feature_extractor = FeatureExtractor(c3d_model).eval()

# ✅ Helper to extract 16-frame clip features
def extract_c3d_features(video_path):
    cap = cv2.VideoCapture(video_path)
    frames = []

    while True:
        ret, frame = cap.read()
        if not ret:
            break
        frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        frame = transform(frame)
        frames.append(frame)

    cap.release()

    if len(frames) < 16:
        return None  # Not enough frames

    clips = []
    for i in range(0, len(frames) - 15, 16):
        clip = torch.stack(frames[i:i+16]).permute(1, 0, 2, 3)  # C, T, H, W
        clips.append(clip.unsqueeze(0))  # Add batch dim

    with torch.no_grad():
        features = [feature_extractor(clip.to(device)).cpu().squeeze().numpy() for clip in clips]

    return np.mean(features, axis=0)  # Average feature for video

# ✅ Load dataset folders
def process_dataset(dataset_path):
    features, labels = [], []
    label_map = {'abuse': 0, 'assault': 1, 'arson': 2, 'arrest': 3}

    for label in os.listdir(dataset_path):
        label_folder = os.path.join(dataset_path, label)
        if not os.path.isdir(label_folder):
            continue
        for video in tqdm(os.listdir(label_folder), desc=f"Processing {label}"):
            video_path = os.path.join(label_folder, video)
            feature = extract_c3d_features(video_path)
            if feature is not None:
                features.append(feature)
                labels.append(label_map[label])

    return np.array(features), np.array(labels)

# ✅ Extract features for training and testing
train_features, train_labels = process_dataset("../dataset/train")
test_features, test_labels = process_dataset("../dataset/test")

# ✅ Save to .pkl
os.makedirs("../features", exist_ok=True)
with open("../features/features.pkl", "wb") as f:
    pickle.dump({
        "train_features": train_features,
        "train_labels": train_labels,
        "test_features": test_features,
        "test_labels": test_labels
    }, f)

print("✅ Features saved to ../features/features.pkl")
```
